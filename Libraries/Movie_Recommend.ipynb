```python
Indian_movies=pd.read_csv("../data/CSV_Datasets/indian movies.csv")
Indian_movies.head(3)
Indian_movies['Language'].unique()
```

```python
import numpy as np
def summ(lst):
    ans=''
    for i in lst:
        ans+=i
    return ans
print(Indian_movies.info())
unq=Indian_movies['Votes'].unique()
print("\n",unq)
Indian_movies['Votes'].replace('-','0',inplace=True)
Indian_movies['Votes'].replace('$5.16M','0',inplace=True)
unq=Indian_movies['Votes'].unique()
print("\n",unq)
# print(type(Indian_movies['Votes'].str.split(',')))
Indian_movies['Votes']=Indian_movies['Votes'].str.split(',')
Indian_movies['Votes']=pd.Series([summ(i) for i in Indian_movies['Votes']])
# Indian_movies['Votes']=Indian_movies['Votes'].apply(lambda x:sum(x))
```

```python
Indian_movies['Votes']=Indian_movies['Votes'].astype(np.int64)
Indian_movies.info()
```
