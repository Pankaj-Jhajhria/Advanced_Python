### BroadCasting

```python
# Broadcasting describes how Numpy treates arrays with different shapes during 
# arithmetic operations.

# The smaller array is 'broadcast' across the larger array so that they have compatible shape.
```

```python
# same shape
a=np.arange(6).reshape(2,3)
b=np.arange(6,12).reshape(2,3)

print(a+b)
```

```python
# diff shape
a=np.arange(6).reshape(2,3)
b=np.arange(3).reshape(1,3)

print(a+b)
```

### BroadCasting Rules

```python
# 1. Make the two arrays have the same number of dimensions
# 2. Make each dimension of the two arrays the same size (if 1 is not there broadcasting can't be done)
```

```python
# if (3,3) (2D) + (3,) (1D)
#  then make (3,) => (1,3) (same dimension (2D))
#  then make (1,3) => (3,3) (same size)     (Braodcasting possible)
```
