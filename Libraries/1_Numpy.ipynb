### Creating Numpy Arrays

```python
import numpy as np
# 1d (vector)
arr =np.array([1,2,2])
print(arr)
print(type(arr))
```

```python
# mulitdimensional
# 2d (matrix)
arr_2d=np.array([[1,2],[4,5]])
print(arr_2d)
```

```python
# 3d (tensors)
arr_3d =np.array([[[1,2,3],[4,5,6]],[[4,5,3],[6,7,4]]])
print(arr_3d)
```

```python
print(np.array([1,2,3],dtype=int))
print(np.array([1,2,3],dtype=float))
print(np.array([1,2,3],dtype=bool))
print(np.array([1,2,3],dtype=complex))
```

```python
arr=np.arange(1,10,3)
arr.reshape(3,1)
```

```python
print(np.ones((3,4)))
print(np.zeros((6,5)))
```

```python
arr =np.random.random((3,4))
print(arr)
```

### Attributes

```python
arr =np.arange(10,dtype=np.int16).reshape(5,2)
```

```python
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.itemsize)   # bytes  (2 bit =>1 bytes)
print(arr.dtype)
```

```python
# astype
arr=np.array([1,2,3],dtype=int)
print(arr.dtype)
# but if it's like age (that doesn't need float or int64)
arr_new=arr.astype(np.int32)
print(arr_new.dtype)
```

### Array Operations

```python
arr=np.arange(12).reshape(3,4)
arr_2=np.arange(12,24).reshape(3,4)
```

```python
print(arr+arr_2)
print("\n",arr+2)
print("\n",arr*2)
print("\n",arr**2)
```

```python
# vector operations
print(arr/arr_2)
```

### Array Functions

```python
arr =np.random.random((3,4))
arr =np.round(arr*100)
arr
```

```python
print(np.max(arr))
print(np.min(arr))
print(np.sum(arr))
print(np.prod(arr))
```

```python
# 0 => col
# 1 => row
print(np.max(arr,axis=0))
```

```python
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr))
```

```python
np.sin(arr)
```

```python
# dot product
dot=np.dot(arr,arr.copy().reshape(4,3))
print(dot)
```

```python
print(np.log(arr))
print()
print(np.exp(arr))
```

```python
# round (nearest integer)
# floor (integer less than)
# ceil (integer greater than)
arr=np.random.random(10)*100
print(arr)
print(np.round(arr))
print(np.floor(arr))
print(np.ceil(arr))
```

### Indexing and Slicing

```python
arr_1d=np.arange(10)
arr_2d=np.arange(10).reshape(2,5)
arr_3d=np.arange(20).reshape(2,2,5)
print("1D Array :",arr_1d)
print("\n2D Array :",arr_2d)
print("\n3D Array :",arr_3d)
```

```python
# Indexing
print(arr_1d[-1])
print(arr_2d[-1])
print(arr_3d[-1])

print()

print(arr_2d[1,1])
print(arr_3d[1,0,3])
```

```python
# Slicing
print(arr_1d[2:5:2])
print(arr_2d[2:5])
print(arr_3d[2:5])

print()

print(arr_2d[:,3:])
print(arr_3d[:,1:,1:4])


# all corners
print("\nall corners :",arr_2d[:,::4])
```

### Iterating

```python
for i in arr_3d:
    print(i)
```

```python
for i in np.nditer(arr_3d):
    print(i)
```

### Reshaping

```python
print(arr_2d)
print()
print("Transpose of a matrix :",np.transpose(arr_2d))
```

```python
# .reshape()
print(arr_3d)
print()
print("Transpose of a tensor :",np.transpose(arr_3d))

# Visualize tensor as 2d matrix plates on top of another making 3d cube
# then transpose :
                #  just watch this cube from left side
arr2_3d=np.arange(27).reshape((3,3,3))
print()
print()
print(arr2_3d)
print("Transpose of a tensor :",np.transpose(arr2_3d))
# print("Transpose of a tensor :",arr2_3d.T)         # easier way to transpose

```
