## Pandas Series

A Pandas Series is like a column in a table , 1D array holding data of any type.

```python
import pandas as pd
import numpy as np
```

#### Creating Series

```python
# Series from lists

country=['India','USA','Nepal']
gdp=[9.8,6.7,4]
print(pd.Series(country,dtype=str,index=['First','Second','Third']))
GDP_Series =pd.Series(gdp,index=country,name='GDP of countries')
print("\n",pd.Series(gdp,index=country,name='GDP of countries'),sep="")
```

```python
# Series from dict   (keys :index)

marks={
    'maths':67,
    'english':89,
    'computer':100
}
marks_series=pd.Series(marks,name='Pankaj ke marks')
print(pd.Series(marks,name='Pankaj ke marks'))
```
