# Encapsulation

Building data + controlling access to it 

##### Preventing direct modifications 
##### ensoring rules through methods

```python
class BankAccount2:
    def __init__(self,balance):
        self._balance = balance
    
    def deposit(self,amount):
        self._balance+=amount
        # print(f"Successfully deposited {amount} to the BankAccount.")
        print(f"+{amount}")
        
    def withdraw(self,amount):
        if self._balance<amount:
            print("Insufficient Balance.")
        else:
            self._balance-=amount
            print(f"-{amount}")
        
    def display_salary(self,boolean):
        if not boolean:
            print("Balance :",self._balance)
        else:
            return self._balance
```

```python
ba_1=BankAccount2(1000)
ba_1.display_salary(False)

# ba_1.balance+=100                  # gives error because balance is protected

ba_1.deposit(567)
ba_1.withdraw(123)
print("Balance after a Month:",ba_1.display_salary(True))
```

```python
from abc import ABC, abstractmethod
# (abstract base class)(abc)

class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):           # Empty method
        pass


class CreditCard(PaymentMethod):       # Inheritance
    def pay(self, amount):
        print(f"Paid {amount} via Credit Card")


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} via UPI")
```

```python
def process_payment(method: PaymentMethod):
    method.pay(100)

process_payment(UPI())
process_payment(CreditCard())
```
