python generators is a simple way of creating iterator

```python
def gen_demo():
    
    yield "First Statement"
    yield "Second Statement"
    yield "Third Statement"
    
gen =gen_demo()          # function is not going to run
print(gen)
print(next(gen))
# print(next(gen))
# print(next(gen))

for i in gen:
    print(i)
```

### Yield vs Return

```python
# yield -=> remembers last yield after each iteration
# return -=> just ends and lost all the memory(local data) after return
```

```python
def my_own_range(start,end):
    for i in range(start,end):
        yield i

for i in my_own_range(15,26):
    print(i)
```

### Generator Expression

```python
gen =(i**2 for i in range(1,101))

for i in gen:
    print(i)
    if i>20:
        break
```
