# Serialization and Deserialization

Serialization  - process of converting python data types to JSON format ( javascript on notation) (Universal text format) (understood by all languages ) (all api's uses JSON format)

Deserialization - JSON to python data types

```python
import json
L=[1,2,3]
D={"name":"Pankaj","age":20}
with open("./data/demo.json","w") as f:
    # json.dump(L,f)
    json.dump(D,f,indent=4)
```

```python
with open("./data/demo.json",'r') as f:
    d=json.load(f)
    print(d)
    print(type(d))
```

tuple stores automatically as list inside json format  (explicitly need to do tuple(L))
