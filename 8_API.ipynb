```python
data=response.json()
print(type(data))
print(len(data))
if type(data) is list:
    print("YES.")
if type(data[0]) is dict:
    print("YES.")
    
def openlist(data):
    for index ,value in enumerate(data):
        if type(value) is list:
            openlist(value)
        if type(value) is dict:
            for key,v in value.items():
                if type(v) is list:
                    openlist(v)
                elif type(v) is dict:
                    print(f"{key} :=> ")
                    for k,v2 in v.items():
                        strr=' '*(len(key)+5)
                        print(strr,end="")
                        print(f"{k} :=> {v2}")
                else:
                    print(f"{key} :=> {v}")    
    
openlist(data)
        
```

```python
import csv
with open("./data/data1.csv","w",newline="") as f:
    writer=csv.writer(f)
    writer.writerow(["id","name","username","email","address","phone","website","company"])
    for user in data:
        keys=list(user.keys())
        num=len(keys)
        lst=[]
        for key in keys:
            if key == "address" or  key =="company":
                add=""
                for k,v in user[key].items():
                    if type(v) is dict:
                        add +="( "+k+" => "
                        for k2 ,v2 in v.items():
                            add+="( "+k2+" => "+v2+" ) "
                        add+=") "
                    else:
                        add+="( "+k+" => "+v+" ) "
                lst.append(add)
            else:
                lst.append(user[key])
        writer.writerow(lst)
```
