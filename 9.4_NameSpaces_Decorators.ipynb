### Scope and LEGB Rule

Scope is a textual region to directly access a namespace (can be visulize at python tutor)  (https://pythontutor.com/visualize.html#mode=display)

LEGB :The interpreter searches for a name from the inside out,according to LEGB. if the interpreter doesn't find the name in any of these locations ,then python raises a NameError exception.

```python
a=2      # Global scope
def temp():
    b=4   # local scope
    print(b)

temp()
print(a)
```

```python
a=2      # Global scope
def temp():
    print(a)         # can view or print Global variable but can't change

temp()
print(a)
```

```python
a=2      # Global scope
def temp():
    global a            # to tell gonna use global variable (now can be modified locally)
    a+=1
    print(a)

temp()
print(a)
```

### Enclosing Scope

```python
print("Hello")   # main scope (global)
def outer():          # enclosing scope(non-local scope)
    def inner():          # local scope
        print('inner function')
    inner()
    print('outer function')
    
outer()
print('main program')
```

```python
def outer():
    a=4
    def inner():
        nonlocal a        # to access and modefy enclosing variables
        a+=1
        print("inner",a)
    inner()
    print("outer",a)

outer() 
```

# Decorators

is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

Can happens Because python functionnnnns are 1st class citizens.  (anythiny can be done with these objects)

```python
# 2 Types :
#         =>Built in decorators           :(@staticmethod,@classmethod,@abstractmethod,@property)
#         =>User defined decorators      
```

```python
# 1st class function
def func():
    print('hello')
a=func
a()
```

```python
del func
# func()      #because it's deleted
```

```python
def modify(func,num):
    return func(num)
def square(num):
    return num**2

modify(square,5)
```

```python
def m_decorator(func,name):
    def wrapper():
        print("**************")
        func(name)
        print("**************")
    return wrapper

def Name(name):
    print(name)

name =input("Enter your name :")
a=m_decorator(Name,name)
a()
```

```python
# Visualize it on python tutor

# even after return of a parent function , a child one can access variables from parent one
#  (Closure)
```

```python
# Another way of syntax
# Go and visualize
def m_decorator(func):
    def wrapper(name):
        print("**************")
        func(name)
        print("**************")
    return wrapper

@decorator
# @m_decorator
def Name(name):
    print(name)

name =input("Enter your name :")
Name(name)
```

```python
import time

def decorator(func):
    def wrapper(*args):
        start =time.time()
        func(*args)
        now=time.time()
        print("\nExecution time of this funciton :",(now-start),"sec.")
    return wrapper

@decorator
def filefiller(name,file_name,iter):
    file_path =f"./data/{file_name}.txt"
    with open(file_path,"w") as f:
        for i in range(iter):
            strr=f"{name} "*(i)+"\n"
            f.write(strr)
    
filefiller("Pankaj","file",1000)
```
