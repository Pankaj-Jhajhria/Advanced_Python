### Scope and LEGB Rule

Scope is a textual region to directly access a namespace (can be visulize at python tutor)  (https://pythontutor.com/visualize.html#mode=display)

LEGB :The interpreter searches for a name from the inside out,according to LEGB. if the interpreter doesn't find the name in any of these locations ,then python raises a NameError exception.

```python
a=2      # Global scope
def temp():
    b=4   # local scope
    print(b)

temp()
print(a)
```

```python
a=2      # Global scope
def temp():
    print(a)         # can view or print Global variable but can't change

temp()
print(a)
```

```python
a=2      # Global scope
def temp():
    global a            # to tell gonna use global variable (now can be modified locally)
    a+=1
    print(a)

temp()
print(a)
```

### Enclosing Scope

```python
print("Hello")   # main scope (global)
def outer():          # enclosing scope(non-local scope)
    def inner():          # local scope
        print('inner function')
    inner()
    print('outer function')
    
outer()
print('main program')
```

```python
def outer():
    a=4
    def inner():
        nonlocal a        # to access and modefy enclosing variables
        a+=1
        print("inner",a)
    inner()
    print("outer",a)

outer() 
```
