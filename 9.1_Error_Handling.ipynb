In python errors are called :- Exceptions


```python
# ValueError (wrong data format) 
# TypeError  (incompatible data types)
# FileNotFoundError  (dataset path issues)
# MemoryError  (large model/ data crashes)
# IndexError  (array indexing mistakes)
# KeyError  (missing dictionary keys)
# ImportError  (missing libraries)
# PermissionError  (no permission for access)
```

```python
y=input("Enter your Marks :")
try :
    x=int(y)
except(ValueError, TypeError):
    print("Conversion Failed.")
else:
    print("Marks :",x)
    
```

```python
# Except as e
y=input("Enter a Integer :")
try :
    x=int(y)
except Exception as e:
    print(f"Failed :{e}")
else:
    print("Age :",x)
```

Custom Exceptions

```python
# raise our own errors

def power(a:int,b:int)->int:
    if(a==0 & b==0):
        raise ValueError("both are becoming 0 ,making power form as (0^0)")
    else:
        return a**b
    
try:
    a,b =map(int ,input("Enter 2 Ingeters :").split())
    p=power(a,b)
except Exception as e:
    print(f"Failed :{e}")
else:
    print(f"Power ({a}^{b}) :{p}")
```
