```python
# try except block

y=input("Enter your Marks :")
try :
    x=int(y)
except(ValueError, TypeError):
    print("Conversion Failed.")
else:                                               # if sure about else work if try worksout
    print("Marks :",x)
finally:
    print("Thanks for using our platform.")
    
```

```python
# Except as e
y=input("Enter a Integer :")
try :
    x=int(y)
except Exception as e:
    print(f"Failed :{e}")
else:
    print("Age :",x)
```

# Custom Exceptions

```python
class SecurityError(Exception):
    def __init__(self,message):
        print(message)
    
class Google:
    def __init__(self,email,password,device):
        self.email=email
        self.password=password
        self.device=device
    
    def login(self,email,password,device):
        if self.device != device:
            raise SecurityError("Your account is just logged in another device.")
        else:
            if self.email ==email and self.password==password:
                print("Succesfully logged in.")
            else:
                print("Invalid Credentials.")
    
try:
    user=Google("lilojhajhria@gmail.com","12345","macIOS")
    user.login("lilojhajhria@gmail.com","12345","victus")
except Exception as e:
    print("Log out from all devices ,if it's not you.")
else:
    print("Save your credentials for next time.")
finally:
    print("Database closed.")
                    
```

```python
# raise our own errors

def power(a:int,b:int)->int:
    if(a==0 & b==0):
        raise ValueError("both are becoming 0 ,making power form as (0^0)")
    else:
        return a**b
    
try:
    a,b =map(int ,input("Enter 2 Ingeters :").split())
    p=power(a,b)
except Exception as e:
    print(f"Failed :{e}")
else:
    print(f"Power ({a}^{b}) :{p}")
```
